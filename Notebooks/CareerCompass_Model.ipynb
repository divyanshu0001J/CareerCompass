```python
branch_keywords = {
   "Architecture": [
    "Architects, Except Landscape and Naval",
    "Landscape Architects"
],
   "Civil": [
    "Civil", "Structural", "Transportation",
    "Water/Wastewater", "Environmental"
],

    "Chemical": [
        "Chemical"
    ],

    "Electrical": [
        "Electrical", "Energy", "Wind Energy",
        "Solar Energy"
    ],

    "Electronics & Communication": [
        "Electronics", "Photonics", "Telecommunications"
    ],

   "Mechanical": [
    "Mechanical", "Automotive", "Robotics",
    "Mechatronics", "Aerospace", "Marine",
    "Industrial", "Manufacturing", "Agricultural"
],

    "Computer Engineering": [
        "Computer Hardware"
    ],

    "Biomedical": [
        "Bioengineer", "Biomedical"
    ]
}

def classify_branch(title):
    for branch, keywords in branch_keywords.items():
        for word in keywords:
            if word.lower() in title.lower():
                return branch

    return "Other"

engineering["Branch"] = engineering["Title"].apply(classify_branch)

print(engineering[["Title", "Branch"]].to_string(index=False))
```

```python
academic_columns = list(academic_mapping.values())

academic_values = {}

for subject in academic_mapping:
    column = academic_mapping[subject]
    academic_values[column] = academic_scores[subject]

academic_differences = career_profiles[
    academic_columns
].sub(pd.Series(academic_values)).abs()

academic_matches = career_profiles[
    ["Title", "Branch"]
].copy()

academic_matches["Academic Match"] = (
    100 - (academic_differences.mean(axis=1) / 4 * 100)
)

academic_matches = academic_matches.sort_values(
    by="Academic Match",
    ascending=False
)

print(academic_matches.head(10).round(2).to_string(index=False))
```

```python
skill_differences = pd.DataFrame(index=career_profiles.index)

for skill in skill_mapping:
    columns = skill_mapping[skill]

    required_level = career_profiles[columns].mean(axis=1)

    student_level = skill_scores[skill]

    skill_differences[skill] = (
        required_level - student_level
    ).abs()

skill_matches = career_profiles[
    ["Title", "Branch"]
].copy()

skill_matches["Skill Match"] = (
    100 - (skill_differences.mean(axis=1) / 4 * 100)
)

skill_matches = skill_matches.sort_values(
    by="Skill Match",
    ascending=False
)

print(skill_matches.head(10).round(2).to_string(index=False))
```

```python
final_matches = career_profiles[
    ["Title", "Branch"]
].copy()

final_matches["Interest Match"] = career_matches[
    "Compatibility"
]

final_matches["Academic Match"] = academic_matches[
    "Academic Match"
]

final_matches["Capability Alignment"] = skill_matches[
    "Skill Match"
]

final_matches["Overall Match"] = (
    final_matches["Interest Match"] * 0.40
    + final_matches["Academic Match"] * 0.30
    + final_matches["Capability Alignment"] * 0.30
)

final_matches = final_matches.sort_values(
    by="Overall Match",
    ascending=False
)

print(
    final_matches[
        ["Title", "Branch", "Interest Match",
         "Academic Match", "Capability Alignment",
         "Overall Match"]
    ].head(10).round(2).to_string(index=False)
)
```
